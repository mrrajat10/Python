# **OOPs**
OOPs is a way of writing programs where we think in terms of objects (real-world things) and their blueprint (class).

# **Components of oops**
* Class: Blue print /plan to construct n number of objects.
* Object : Physical existence of class/AN INSTANCE
* Variables: Properties of object
* methods : Function within a class which are referred as behaviour of object

**Class**

Think of a class like the architect’s blueprint of a house.
It doesn’t build anything by itself, but it defines what a house should have (like number of rooms, color, design).

**Object**

An object is the actual house built from that blueprint.
From the same plan, you can build 10 different houses — one blue, one red, one big, one small.
So, a class is the design, and an object is the real thing standing on the ground.

**Variables (Properties)**

These are the features of the house.
For example:

Color of the house

Number of rooms

Number of floors
They describe what the house is.

**Methods (Behaviors)**

These are the actions the house can do or allow.
For example:

“Open the door”

“Paint the house”

“Switch on the lights”
They describe what the house can do.

# Instance variable
Variable values are different for different objects

Usually defined inside the constructor (__init__ method in Python) or other methods, and are prefixed with self in Python.

```python
class calci:
  def __init__(self , a , b):
    self.a = a
    self.b = b
  def add(self):
    return self.a + self.b
c1 = calci(a = 10 , b = 20)
c2 = calci(a = 100 , b = 200)
```

```python
print(c1.a,c2.a)
```

```python
print(c1.add())
print(c2.add())
```

# Inheritence
In Python OOP, inheritance allows one class (child/subclass) to acquire the properties and methods of another class (parent/superclass). This promotes reusability and helps build hierarchical relationships.

#Single Inheritance

A child class inherits from one parent class.

Single inheritance is when a child class (derived class) inherits properties and behaviors (attributes and methods) from only one parent class (base class).

**A company wants to calculate salaries for its employees.**

- Create a base class Employee with an attribute base_salary.
- Create a child class PartTimeEmployee that inherits Employee.
- Part-time employees earn 60% of base salary.
- Create an object of PartTimeEmployee with salary 20000 and calculate final salary.

```python
class Employee: #parent class
  def __init__(self,base_salary):
    self.base_salary=base_salary
class PartTimeEmployee(Employee): #child
  def salary(self):
    return self.base_salary * 0.6
emp1 = PartTimeEmployee(20000)
print(emp1.salary())
```

# Multiple Inheritance
Multiple Inheritance means a child class inherits from more than one parent class.
This allows the child to combine attributes and methods from multiple parents.

**Write a Python program using multiple inheritance where:**

- Employee class stores name and salary.
- BonusProvider class provides bonus = 5000.
- Manager class inherits from both and calculates total salary.

Create an object of Manager with base salary 30000 and display final salary.

```python
class Employee: # Parent 1
  def __init__(self,name,salary):
    self.name=name
    self.salary=salary
  def get_salary(self):
    return f"salary before bonus {self.salary}"
class BonusProvider: #parent2
  def get_bonus(self):
    return 5000
class Manager(Employee , BonusProvider): #child
  def total_salary(self):
    return f"The total salary of {self.name} is {self.salary + self.get_bonus()}"
emp1 = Manager('John',40000)
print(emp1.get_salary())
print(emp1.total_salary())

```

#**Multilevel Inheritence**
Multilevel Inheritance means that a class inherits from a child class, which itself inherits from another parent class.
It forms a chain of inheritance (like a family tree across generations).

**🔹 Real-World Analogy**

**👴 Grandfather → 👨 Father → 👦 Child**

Father inherits traits from Grandfather.

Child inherits traits from both Father and Grandfather.

In a company, salary is calculated in levels of roles:
**Parent → Employee**
Stores base salary.

**Child → FullTimeEmployee**
Adds a fixed bonus of 5000.

**Grandchild → Manager**
Adds an extra allowance of 10000.

This creates a chain of inheritance:

**Employee → FullTimeEmployee → Manager**

Write a Python program using multilevel inheritance where:

- Employee class stores base salary.
- FullTimeEmployee adds a bonus.
- Manager adds an extra allowance.

Create a Manager with base salary = 20000 and calculate final salary.

```python
# grand parent
class Employee:
  def __init__(self,salary):
    self.salary=salary
  def get_salary(self):
    return self.salary
# parent
class FullTimeEmployee(Employee):
  def add_bonus(self):
    return self.salary + 5000
# child
class Manager(FullTimeEmployee):
  def updated_salary(self ):
    return self.add_bonus() + 10000
obj = Manager(40000)
```

```python
# actual salary
obj.get_salary()
```

```python
# salary after bonus
obj.add_bonus()
```

```python
# salary after bonus and after extra allowance
obj.updated_salary()
```

```python
# grand parent
class Employee:
  def __init__(self,salary):
    self.salary=salary
  def get_salary(self):
    return self.salary
# parent
class FullTimeEmployee(Employee):
  def add_bonus(self,bonus):
    return self.salary + bonus
# child
class Manager(FullTimeEmployee):
  def updated_salary(self,bonus, extra_allowance ):
    return self.add_bonus(bonus) + extra_allowance
obj = Manager(40000)
```

```python
obj.get_salary()
```

```python
obj.add_bonus(5000)
```

```python
obj.updated_salary(5000,10000)
```
